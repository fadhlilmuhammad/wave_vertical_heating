```python
yyyymmdd = "19810201"
yyyy = yyyymmdd[0:4]
ens = "e01"

folder_access = {}
ds_access = {} 

for var in ["u", "v", "w", "q"]:
    folder_access[var] = Path(f"/home/563/fm6730/localrepo/wave_vertical_heating/data/temp/wave_vertical_heating/{var}/access-s2")
    
    ds_access[var] = xr.open_dataset(os.path.join(folder_access[var],f"da_{var}_{yyyymmdd}_{ens}_2-5deg.nc"))
```
